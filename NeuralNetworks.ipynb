    <h1 style="font-size: 72px; font-weight: 700; margin-left: 10px;">Let’s design a 2-layer MLP</h1>
    <hr style="width: 50%; margin-left: -0%; margin-top: -0%;" />
    <div style="display: flex; flex-direction: row-reverse; justify-content: center;">
        <div style="width: 50%; margin-top: 1%;">
            <img src="images/mlp1.png" alt="Multi-layer Perceptron" />
        </div>
        <div style="width: 40%; margin-left: 8%;">
            <ul style="font-size: 30px;">
                <li>
                    Layer-2: \(\hat{y} = g(v \cdot h)\)
                </li>
                <li>
                    Layer-1: \(h_i = f(w_i \cdot x)\)
                </li>
                <li>
                    \(f, g\) are called activation/link functions
                    <ul style="list-style: none; margin-left: 40px; margin-top: 15px;">
                        <li style="margin-bottom: 20px;">
                            ▸ <span style="color: #A62B17;">hyperbolic tangent (\(\mathit{tanh}\))</span>
                        </li>
                        <li style="margin-bottom: 20px;">
                            ▸ <span style="color: #3274B5;">Sigmoid \(\sigma = \frac{1}{e^{-x}+1}\)</span>
                        </li>
                        <li style="margin-bottom: 20px;">
                            ▸ Rectified Linear Unit (ReLU)
                        </li>
                        <li style="margin-bottom: 20px;">
                            ▸ Identity (usually not in Layer-1)
                        </li>
                    </ul>
                </li>
                <li>
                    Set \(f = \mathit{tanh}, g = \mathit{identity}\)
                </li>
                <li>
                    \(\hat{y} = \sum_{i} v_i \mathit{tanh}(w_i \cdot x) = v \cdot \mathit{tanh}(W \cdot x)\)
                </li>
            </ul>
        </div>

```python
import matplotlib.pyplot as plt
import numpy as np

# XOR data (±1 encoding, as in the table on the XOR slide)
X = np.array([[-1, -1], [-1, 1], [1, -1], [1, 1]])
y_true = np.array([-1, 1, 1, -1])

# Hand-picked weights that solve XOR with tanh hidden units
#   hidden unit 1 ≈ OR : h1 = tanh(5·x1 + 5·x2 + 5)  is -1 only for (-1,-1)
#   hidden unit 2 ≈ AND: h2 = tanh(5·x1 + 5·x2 - 5)  is +1 only for (1,1)
#   output: y_hat = h1 - h2 - 1   ("OR and not AND" = XOR)
W1 = np.array([[5, 5], [5, 5]])  # 2x2: input to hidden (Input -> Layer 1), one row per hidden unit
b1 = np.array([5, -5])           # bias for hidden layer
v = np.array([1, -1])            # hidden to output weights (Layer 1 -> Layer 2)
b2 = -1                          # output bias

# Forward pass
def forward(X, W1, b1, v, b2):
    z1 = X @ W1.T + b1  # hidden layer (Layer 1) pre-activation
    h1 = np.tanh(z1)    # hidden layer (Layer 1) activation
    z2 = h1 @ v + b2    # output (Layer 2) pre-activation
    y_hat = z2          # identity output, as on the slides; predict sign(y_hat)
    return z1, h1, z2, y_hat

# Call the forward method to perform the forward pass
z1, h1, z2, y_hat = forward(X, W1, b1, v, b2)
for x, h, out, t in zip(X, h1, y_hat, y_true):
    print(f"x = ({x[0]:+d},{x[1]:+d}) -> h = ({h[0]:+.3f}, {h[1]:+.3f}) -> y_hat = {out:+.3f} -> sign {int(np.sign(out)):+d}   target {t:+d}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create visualization with just 2 plots
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Input space (XOR data)
ax = axes[0]
colors = ['blue' if y == 1 else 'red' for y in y_true]  # +1 blue, -1 red
ax.scatter(X[:, 0], X[:, 1], c=colors, s=300, edgecolors='black', linewidth=2)
for i in range(len(X)):
    ax.text(X[i, 0], X[i, 1], f'({X[i, 0]},{X[i, 1]})', ha='center', va='center', 
            fontsize=8, color='white', fontweight='bold')
ax.set_xlabel('$x_1$', fontsize=14)
ax.set_ylabel('$x_2$', fontsize=14)
ax.set_title('Input Space (XOR)', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)

# Plot 2: Single decision boundary attempt (fails for XOR)
ax = axes[1]
ax.scatter(X[:, 0], X[:, 1], c=colors, s=300, edgecolors='black', linewidth=2)
for i in range(len(X)):
    ax.text(X[i, 0], X[i, 1], f'({X[i, 0]},{X[i, 1]})', ha='center', va='center', 
            fontsize=8, color='white', fontweight='bold')

# Try to draw a single linear decision boundary (it will fail to separate XOR)
xx = np.linspace(-1.5, 1.5, 100)
yy = np.linspace(-1.5, 1.5, 100)
XX, YY = np.meshgrid(xx, yy)

# Example decision boundary: x1 + x2 = 0 (or any other line you choose)
Z = XX + YY  # This will fail to properly separate XOR

ax.contour(XX, YY, Z, levels=[0], colors='green', linewidths=3, linestyles='--')
ax.contourf(XX, YY, Z, levels=[-100, 0, 100], colors=['lightcoral', 'lightblue'], alpha=0.3)

ax.set_xlabel('$x_1$', fontsize=14)
ax.set_ylabel('$x_2$', fontsize=14)
ax.set_title('Single Decision Boundary\n(Cannot Solve XOR!)', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)

plt.tight_layout()
plt.show()
```

```python
# How the hidden layer solves XOR: look at the data the way the output unit sees it
from collections import defaultdict

z1, h, z2, y_hat = forward(X, W1, b1, v, b2)
pos, neg = y_true == 1, y_true == -1
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Left: input space, the two hidden units' lines, and the network's decision regions
ax = axes[0]
g = np.linspace(-1.6, 1.6, 300)
G1, G2 = np.meshgrid(g, g)
out = forward(np.c_[G1.ravel(), G2.ravel()], W1, b1, v, b2)[3].reshape(G1.shape)
ax.contourf(G1, G2, out, levels=[-100, 0, 100], colors=['lightcoral', 'lightblue'], alpha=0.35)
for i, (color, name) in enumerate([('teal', '$h_1 = 0$ (≈ OR)'), ('darkorange', '$h_2 = 0$ (≈ AND)')]):
    ax.plot(g, -(W1[i, 0] * g + b1[i]) / W1[i, 1], color=color, lw=2.5, label=name)  # w_i·x + b_i = 0
ax.scatter(*X[pos].T, c='blue', s=250, marker='o', edgecolors='black', zorder=3, label='+1')
ax.scatter(*X[neg].T, c='red', s=250, marker='s', edgecolors='black', zorder=3, label='−1')
for p in X:
    ax.annotate(f'({p[0]},{p[1]})', p, textcoords='offset points', xytext=(12, 8), fontsize=11)
ax.set(xlim=(-1.6, 1.6), ylim=(-1.6, 1.6), xlabel='$x_1$', ylabel='$x_2$')
ax.set_title('Input space: two hidden-unit lines\ngive one non-linear boundary', fontsize=13, fontweight='bold')
ax.set_aspect('equal')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.12), ncol=4, fontsize=10)

# Right: hidden space (h1, h2), where a single line now separates the classes
ax = axes[1]
H1, H2 = np.meshgrid(np.linspace(-1.4, 1.4, 200), np.linspace(-1.4, 1.4, 200))
Zh = v[0] * H1 + v[1] * H2 + b2
ax.contourf(H1, H2, Zh, levels=[-100, 0, 100], colors=['lightcoral', 'lightblue'], alpha=0.35)
ax.contour(H1, H2, Zh, levels=[0], colors='black', linewidths=2)
ax.scatter(*h[pos].T, c='blue', s=250, marker='o', edgecolors='black', zorder=3)
ax.scatter(*h[neg].T, c='red', s=250, marker='s', edgecolors='black', zorder=3)
groups = defaultdict(list)                       # (1,-1) and (-1,1) land on the same hidden point
for p, hp in zip(X, h):
    groups[tuple(np.round(hp, 2))].append(f'({p[0]},{p[1]})')
for hp, names in groups.items():
    below = hp[1] < 0 and hp[0] > 0
    ax.annotate(' & '.join(names), hp, textcoords='offset points', xytext=(10, -32) if below else (-16, 12),
                ha='right', fontsize=11)
ax.text(0.3, -0.55, '$h_1 - h_2 - 1 = 0$', fontsize=12, rotation=45, ha='center', va='bottom',
        rotation_mode='anchor')
ax.set(xlim=(-1.4, 1.4), ylim=(-1.4, 1.4), xlabel='$h_1$', ylabel='$h_2$')
ax.set_title('Hidden space: the classes are now\nlinearly separable', fontsize=13, fontweight='bold')
ax.set_aspect('equal')

plt.tight_layout()
plt.show()
print("The output unit only sees (h1, h2). In that space one straight line separates the classes.")
print("Try it: change W1, b1, v or b2 in the previous cell and re-run both cells.")
```

    <h1 style="font-size: 72px; font-weight: 700; margin-left: 10px; color: #333;">Recap: General Gradient Descent Algorithm</h1>
    <hr style="width: 50%; margin-left: -0%; margin-top: 3%;" />
    <div style="font-size: 20px;">
         \[
            \begin{array}{ll}
        \hline
            \textbf{Algorithm: } & \text{GRADIENT DESCENT}(L(\mathbf{w}, b; \mathbf{D}), \mathbf{D}, maxIter, \eta_1, \ldots, \eta_k) \\
            \hline
            1: & \mathbf{w}^{(0)} \leftarrow \langle 0, 0, \ldots, 0 \rangle, b^{(0)} \leftarrow 0 \quad \text{// initialise variables we are optimising (weights vector and bias)} \\
            2: & \textbf{for } k = 1 \ldots maxIter \textbf{ do} \\
            3: & \quad \mathbf{g}^{(k)} \leftarrow \nabla_{w,b} L(w, b; \mathbf{D})|_{w,b^{(k-1)}} \quad \text{// compute gradient (slope of the loss) at current location} \\
            4: & \quad \mathbf{w}^{(k)} \leftarrow \mathbf{w}^{(k-1)} - \eta^{(k)} \mathbf{g}^{(k)} \quad \text{// take a step down the gradient; update weights in the direction that reduces loss} \\
            5: & \quad b^{(k)} \leftarrow b^{(k-1)} - \eta \frac{\partial L^{(k)}}{\partial b} \text{// update bias in the direction that reduces loss} \\
            6: & \textbf{end for} \\
            7: & \textbf{return}
            \end{array}
            \]
    </div>
    <div style="display: flex; flex-flow: column; font-size: 16px; margin-top: 3%;">
        <div style="width: 40%; margin-right: 9%;">
            \[
                \begin{array}{ll}
                \textbf{Greek letters:} & \\
                \eta \text{ (eta)} & \text{learning rate (step size for updates)} \\
                \nabla \text{ (nabla)} & \text{gradient operator (takes derivative)} \\
                \partial & \text{partial derivative symbol} \\
                \\
                \textbf{Notation:} & \\
                w^{(k)} & \text{value of } w \text{ at iteration } k \\
                w^{(k-1)} & \text{value of } w \text{ at previous iteration} \\
                \langle 0, 0, \ldots, 0 \rangle & \text{vector of zeros} \\
                \leftarrow & \text{assignment (set equal to)} \\
                |_{w,b^{(k-1)}} & \text{evaluated at previous values}
                \end{array}
                \]
        </div>
        <div style="width: 40%; margin-left: 1%;">
            \[
                \begin{array}{ll}
                \textbf{Variables:} & \\
                \mathbf{w} & \text{weights (parameters we're optimising)} \\
                b & \text{bias (another parameter we're optimising)} \\
                L & \text{loss function (measures prediction error)} \\
                \mathbf{D} & \text{dataset (training data)} \\
                \mathbf{g} & \text{gradient (direction to update parameters)} \\
                k & \text{iteration number (current step)} \\
                n & \text{index for data points} \\
                \end{array}
                \]
        </div>
    </div>

    <h1 style="font-size: 72px; font-weight: 700; margin-left: 10px; color: #333;">Two-layer NN: Gradient of output layer weights \(v\)</h1>
    <hr style="width: 50%; margin-left: -0%; margin-top: -0%;" />
    <div style="display: flex; flex-flow: column;">
        <div style="width: 60%; margin-right: 5%; font-size: 30px;">
            The Loss Function:
            <span style="font-size: 22px;">
            \[
            L(\mathbf{W}, v) = \sum_{n} \frac{1}{2}\left(y_n - \sum_{i} v_i \cdot f(\mathbf{w}_i \cdot \mathbf{x}_n)\right)^2
            \]
            </span>
            Breaking this down:
            <ul>
                <li>For each training example n, we compute the prediction: \(\hat{y}_n = \sum_i v_i \cdot f(w_i \cdot x_n)\)</li>
                <li>The error for example n is: \(e_n = y_n − \hat{y}_n\)</li>
                <li>The loss is: \(L = \frac{1}{2}e_n^2\)</li>
            </ul>
            We want to find: How does the loss change when we change the output weight \(v_i\)?<br/>
            That is: \(\frac{\partial L}{\partial v_i}\)
        </div>

    <h1 style="font-size: 72px; font-weight: 700; margin-left: 10px; color: #333;">Two-layer NN: Gradient of output layer weights \(v\)</h1>
    <hr style="width: 50%; margin-left: -0%; margin-top: -0%;" />
    <div style="display: flex; flex-flow: column;">
        <div style="width: 60%; margin-right: 5%; font-size: 30px;">
            The Math of \(\frac{\partial L}{\partial v_i}\) (Chain Rule):
            <span style="font-size: 22px;">
            \[
            \frac{\partial L}{\partial v_i} = \frac{\partial L}{\partial g} \cdot \frac{\partial g}{\partial v_i}
            \]
            </span>
            <b>Part 1</b>: How does loss change with the prediction?<br/>
            \[
                L = \frac{1}{2}(y_n - g)^2 = \frac{1}{2}e_n^2
            \]
            \[
                \frac{\partial L}{\partial g} = \frac{\partial L}{\partial e} \cdot \frac{\partial e}{\partial g} = (y_n - g) \cdot -1 = -(y_n - g) = -e_n
            \]
            We use \(g\) to represent the network output (the prediction)
            <br/>
            <br/>
            <b>Part 2</b>: How does the prediction change with weight \(v_i\)?
                \[
                g = identity\left(\sum_i v_i \cdot h_{i,n}\right) = v_1 \cdot h_{1,n} + v_2 \cdot h_{2,n} + ... + v_i \cdot h_{i,n}
                \]
                \[
                \frac{\partial g}{\partial v_i} = h_{i,n}
                \]
                (The prediction changes by exactly \(h_{i,n}\) when we change \(v_i\) by 1)

                L(\mathbf{W}, v) = \frac{1}{2}\left(y - \sum_{i} v_i \cdot f(\mathbf{w}_i \cdot \mathbf{x})\right)^2 = \frac{1}{2}e^2

            <br/>

            \[
                \frac{\partial L}{\partial \mathbf{w}_i} = \frac{\partial L}{\partial f_i} \cdot \frac{\partial f_i}{\partial \mathbf{w}_i}
            \]
            In simple words: to find how the loss changes with weight \(w_i\), we need to go through the activation \(f_i\) of the hidden unit \(h_i\)

    <h1 style="font-size: 72px; font-weight: 700; margin-left: 10px; color: #333;">Backprop for 2-layer Network</h1>
    <hr style="width: 50%; margin-left: -0%; margin-top: -0%;" />
    <div style="display: flex; flex-flow: column; font-size: 20px;">
        <div style="width: 60%; margin-right: 5%;">
            \[
                \begin{array}{ll}
                \textbf{Algorithm 25 } & \textbf{TWO LAYER NETWORK TRAIN}(\mathbf{D}, \eta, K, MaxIter) \\
                \hline
                1: & \mathbf{W} \leftarrow D \times K \text{ matrix of small random values} \quad \text{// initialize input layer weights} \\
                2: & v \leftarrow K\text{-vector of small random values} \quad \text{// initialize output layer weights} \\
                3: & \textbf{for } iter = 1 \ldots MaxIter \textbf{ do} \\
                4: & \quad \mathbf{G} \leftarrow D \times K \text{ matrix of zeros} \quad \text{// initialize input layer gradient} \\
                5: & \quad g \leftarrow K\text{-vector of zeros} \quad \text{// initialize output layer gradient} \\
                6: & \quad \textbf{for all } (x,y) \in \mathbf{D} \textbf{ do} \\
                7: & \quad\quad \textbf{for } i = 1 \text{ to } K \textbf{ do} \\
                8: & \quad\quad\quad a_i \leftarrow w_i \cdot x \\
                9: & \quad\quad\quad h_i \leftarrow \tanh(a_i) \quad \text{// compute activation of hidden unit } i \\
                10: & \quad\quad \textbf{end for} \\
                11: & \quad\quad \hat{y} \leftarrow v \cdot h \quad \text{// compute output unit} \\
                12: & \quad\quad e \leftarrow y - \hat{y} \quad \text{// compute error} \\
                13: & \quad\quad g \leftarrow g - eh \quad \text{// update gradient for output layer} \\
                14: & \quad\quad \textbf{for } i = 1 \text{ to } K \textbf{ do} \\
                15: & \quad\quad\quad \mathbf{G}_i \leftarrow \mathbf{G}_i - ev_i(1 - \tanh^2(a_i))x \quad \text{// update gradient for input layer} \\
                16: & \quad\quad \textbf{end for} \\
                17: & \quad \textbf{end for} \\
                18: & \quad \mathbf{W} \leftarrow \mathbf{W} - \eta\mathbf{G} \quad \text{// update input layer weights} \\
                19: & \quad v \leftarrow v - \eta g \quad \text{// update output layer weights} \\
                20: & \textbf{end for} \\
                21: & \textbf{return } \mathbf{W}, v
                \end{array}
            \]
            <p style="font-size: 18px;">Biases: append a constant 1 to \(x\) and to \(h\), so each \(w_i\) and \(v\) gets one extra weight (the same trick as \(x_0 = 1\) in linear regression). The code later in this notebook does this.</p>

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

plt.style.use('seaborn-v0_8')
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 14

# Customer Transaction Data: the shared Week 7 file (data/customer_transactions.csv)
# The network below uses three of its four features, so we can draw it in 3-D.
FEATURES = ['Music Download?', 'Music Streaming?', 'Online Games']
df = pd.read_csv('data/customer_transactions.csv')[['Transaction'] + FEATURES + ['Output']]
print("=" * 80)
print("CUSTOMER TRANSACTION DATA")
print("=" * 80)
print(df.to_string(index=False))
```

```python
# Convert Yes/No to 1/0 and Buys/Cancels to 1/0 (for regression-style loss)
df_vec = df.copy()
df_vec[FEATURES] = (df[FEATURES] == 'Yes').astype(int)
df_vec['Output'] = (df['Output'] == 'Buys').astype(int)  # Buys=1, Cancels=0
print("\n" + "=" * 80)
print("VECTORIZED DATA")
print("=" * 80)
print("Features:")
print("  • Yes → 1")
print("  • No → 0")
print("\nClass (Output):")
print("  • Buys → 1")
print("  • Cancels → 0")
print()
print(df_vec.to_string(index=False))
```

```python
# Create feature matrix X and labels y
X = df_vec[FEATURES].values
y = df_vec['Output'].values

print(f"\nFeature matrix X shape: {X.shape}")
print(f"Labels y shape: {y.shape}")
print()

# Network parameters
D = X.shape[1]  # Number of input features (3)
K = 2           # Number of hidden units
eta = 0.1      # Learning rate
MaxIter = 300   # Maximum iterations
USE_BIAS = True # bias trick: append a constant 1 to x (hidden biases) and to h (output bias)

print("=" * 80)
print("TWO-LAYER NETWORK TRAINING")
print("=" * 80)
print(f"Network architecture:")
print(f"  • Input features (D): {D}")
print(f"  • Hidden units (K): {K}")
print(f"  • Output units: 1")
print(f"  • Learning rate (η): {eta}")
print(f"  • Max iterations: {MaxIter}")
print(f"  • Bias terms: {USE_BIAS}")
print()
```

```python
# Initialize weights
np.random.seed(42)  # For reproducibility
X_in = np.column_stack([X, np.ones(len(X))]) if USE_BIAS else X  # append the constant 1 for the bias
W = np.random.randn(X_in.shape[1], K) * 0.1  # (D+1)×K matrix: the last row holds the hidden biases
v = np.random.randn(K + int(USE_BIAS)) * 0.1  # K-vector, plus one entry for the output bias

print("Initial weights:")
print(f"  W (input to hidden) =\n{W}")
print(f"  v (hidden to output) = {v}")
print()

# Training history
history = {
    'loss': [],
    'W': [],
    'v': [],
    'predictions': []
}
```

```python
# Training loop
print("=" * 80)
print("TRAINING PROGRESS")
print("=" * 80)

for iteration in range(MaxIter):
    # Initialize gradients
    G = np.zeros_like(W)  # D×K matrix for hidden layer
    g = np.zeros_like(v)  # K-vector for output layer
    
    total_loss = 0
    predictions = []
    
    # Loop over all training examples
    for idx in range(len(X)):
        x = X_in[idx]   # includes the constant 1 when USE_BIAS
        y_true = y[idx]
        
        # ===== FORWARD PASS =====
        # Compute hidden layer activations
        a = np.dot(W.T, x)  # K-vector: weighted inputs to hidden units
        h = np.tanh(a)      # K-vector: hidden activations
        h_out = np.append(h, 1) if USE_BIAS else h  # constant 1 for the output bias
        
        # Compute output
        y_pred = np.dot(v, h_out)  # Scalar: network output
        
        # Compute error and loss
        e = y_true - y_pred
        loss = 0.5 * e**2
        total_loss += loss
        predictions.append(y_pred)
        
        # ===== BACKWARD PASS =====
        # Update output layer gradient
        g = g - e * h_out  # Accumulate: ∇v = -e·h
        
        # Update hidden layer gradients
        for i in range(K):
            # Gradient for hidden unit i
            grad_i = -e * v[i] * (1 - np.tanh(a[i])**2) * x
            G[:, i] = G[:, i] + grad_i
    
    # Update weights (after processing all examples-batch update)
    W = W - eta * G
    v = v - eta * g
    
    # Store history
    history['loss'].append(total_loss)
    history['W'].append(W.copy())
    history['v'].append(v.copy())
    history['predictions'].append(predictions.copy())
    
    # Print progress every 10 iterations
    if iteration % 10 == 0 or iteration == MaxIter - 1:
        avg_loss = total_loss / len(X)
        print(f"Iteration {iteration:3d}: Total Loss = {total_loss:.4f}, Avg Loss = {avg_loss:.4f}")

print()
print("=" * 80)
print("TRAINING COMPLETE")
print("=" * 80)
print(f"Final weights:")
print(f"  W (input to hidden) =\n{W}")
print(f"  v (hidden to output) = {v}")
print()
```

```python
# Final predictions
print("=" * 80)
print("FINAL PREDICTIONS")
print("=" * 80)
print(f"{'Transaction':<12} {'True':<6} {'Predicted':<10} {'Rounded':<8} {'Correct?'}")
print("-" * 60)

for idx in range(len(X)):
    x = X_in[idx]
    y_true = y[idx]
    
    # Forward pass
    a = np.dot(W.T, x)
    h = np.tanh(a)
    y_pred = np.dot(v, np.append(h, 1) if USE_BIAS else h)
    y_rounded = 1 if y_pred > 0.5 else 0
    correct = "✓" if y_rounded == y_true else "✗"
    
    print(f"{df['Transaction'][idx]:<12} {y_true:<6} {y_pred:<10.4f} {y_rounded:<8} {correct}")

# Calculate accuracy
def predict(x_in):
    h = np.tanh(W.T @ x_in)
    return v @ (np.append(h, 1) if USE_BIAS else h)

correct_predictions = sum(1 for idx in range(len(X)) if (1 if predict(X_in[idx]) > 0.5 else 0) == y[idx])
accuracy = correct_predictions / len(X) * 100
print()
print(f"Training Accuracy: {accuracy:.1f}%")
print("Note: T1 and T8 have the same three features but different outputs, so no model can get both right.")
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
import warnings

warnings.filterwarnings('ignore', message='Glyph.*missing from font.*')

def network_output(x, W, v):
    """Compute network output for input x"""
    if W.shape[0] == len(x) + 1:   # bias trick: constant 1 appended to the input
        x = np.append(x, 1)
    h = np.tanh(np.dot(W.T, x))
    if len(v) == len(h) + 1:       # output bias
        h = np.append(h, 1)
    return np.dot(v, h)

def create_decision_surface(W, v, resolution=25):
    """Create a mesh representing the decision boundary (output = 0.5)"""
    x_range = np.linspace(-0.2, 1.2, resolution)
    y_range = np.linspace(-0.2, 1.2, resolution)
    z_range = np.linspace(-0.2, 1.2, resolution)
    
    grid_output = np.zeros((resolution, resolution, resolution))
    
    for i, x_val in enumerate(x_range):
        for j, y_val in enumerate(y_range):
            for k, z_val in enumerate(z_range):
                x_test = np.array([x_val, y_val, z_val])
                grid_output[i, j, k] = network_output(x_test, W, v)
    
    return x_range, y_range, z_range, grid_output

def plot_network_decision_surface(ax, W, v, X, y, df, title):
    """Plot the decision boundary as a continuous surface"""
    
    # Create decision surface
    x_range, y_range, z_range, grid_output = create_decision_surface(W, v, resolution=25)
    
    print(f"Grid output range: [{grid_output.min():.3f}, {grid_output.max():.3f}]")
    
    surface_plotted = False
    
    # Try marching cubes if 0.5 is within the range
    if grid_output.min() < 0.5 < grid_output.max():
        try:
            from skimage import measure
            
            # Extract isosurface at output = 0.5
            verts, faces, _, _ = measure.marching_cubes(grid_output, level=0.5)
            
            # Scale vertices to actual coordinate range
            verts[:, 0] = verts[:, 0] / (len(x_range) - 1) * (x_range[-1] - x_range[0]) + x_range[0]
            verts[:, 1] = verts[:, 1] / (len(y_range) - 1) * (y_range[-1] - y_range[0]) + y_range[0]
            verts[:, 2] = verts[:, 2] / (len(z_range) - 1) * (z_range[-1] - z_range[0]) + z_range[0]
            
            # Plot the mesh
            ax.plot_trisurf(verts[:, 0], verts[:, 1], faces, verts[:, 2],
                           color='cyan', alpha=0.4, edgecolor='blue', linewidth=0.2)
            surface_plotted = True
            
        except ImportError:
            print("scikit-image not available")
    
    # Fallback: use slice method
    if not surface_plotted:
        print("Using slice visualization method...")
        n_slices = 10
        
        for z_val in np.linspace(0, 1, n_slices):
            X1, X2 = np.meshgrid(x_range, y_range)
            Z_out = np.zeros_like(X1)
            
            for i in range(X1.shape[0]):
                for j in range(X1.shape[1]):
                    x_test = np.array([X1[i, j], X2[i, j], z_val])
                    Z_out[i, j] = network_output(x_test, W, v)
            
            # Check if 0.5 is in range for this slice
            if Z_out.min() < 0.5 < Z_out.max():
                ax.contour(X1, X2, Z_out, levels=[0.5], 
                          colors='blue', linewidths=2, alpha=0.6, offset=z_val)
        
        # Also do Y slices
        for y_val in np.linspace(0, 1, n_slices):
            X1, X3 = np.meshgrid(x_range, z_range)
            Z_out = np.zeros_like(X1)
            
            for i in range(X1.shape[0]):
                for j in range(X1.shape[1]):
                    x_test = np.array([X1[i, j], y_val, X3[i, j]])
                    Z_out[i, j] = network_output(x_test, W, v)
            
            if Z_out.min() < 0.5 < Z_out.max():
                ax.contour(X1, Z_out, X3, levels=[0.5], 
                          colors='blue', linewidths=2, alpha=0.4, offset=y_val)
    
    # Plot data points
    for idx in range(len(X)):
        color = 'green' if y[idx] == 1 else 'red'
        ax.scatter(*X[idx], color=color, s=250, alpha=0.9, 
                  edgecolors='black', linewidth=2, zorder=10)
        ax.text(X[idx, 0], X[idx, 1], X[idx, 2] + 0.15, df['Transaction'][idx], 
                fontsize=9, ha='center', weight='bold')
    
    # Labels and formatting
    ax.set_xlabel('Music Download', fontsize=11, weight='bold')
    ax.set_ylabel('Music Streaming', fontsize=11, weight='bold')
    ax.set_zlabel('Online Games', fontsize=11, weight='bold')
    ax.set_title(title, fontsize=12, weight='bold', pad=10)
    
    ax.set_xlim([-0.3, 1.3])
    ax.set_ylim([-0.3, 1.3])
    ax.set_zlim([-0.3, 1.3])
    
    # Add legend
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor='green', label='Buys (y=1)'),
        Patch(facecolor='red', label='Cancels (y=0)'),
        Patch(facecolor='cyan' if surface_plotted else 'blue', 
              alpha=0.4, label='Decision Boundary')
    ]
    ax.legend(handles=legend_elements, loc='upper left', fontsize=9)
    
    ax.view_init(elev=20, azim=45)

# Create visualization
fig = plt.figure(figsize=(18, 6))

# Initial state
W_init = history['W'][0]
v_init = history['v'][0]
ax1 = fig.add_subplot(131, projection='3d')
plot_network_decision_surface(ax1, W_init, v_init, X, y, df, 
                              f'Initial State (Iteration 0)\nRandom Weights')

# Mid-training
mid_iter = len(history['W']) // 2
W_mid = history['W'][mid_iter]
v_mid = history['v'][mid_iter]
ax2 = fig.add_subplot(132, projection='3d')
plot_network_decision_surface(ax2, W_mid, v_mid, X, y, df, 
                              f'Mid-Training (Iteration {mid_iter})\nLoss: {history["loss"][mid_iter]:.3f}')

# Final state
W_final = history['W'][-1]
v_final = history['v'][-1]
ax3 = fig.add_subplot(133, projection='3d')
plot_network_decision_surface(ax3, W_final, v_final, X, y, df, 
                              f'Final State (Iteration {MaxIter-1})\nLoss: {history["loss"][-1]:.3f}')

plt.tight_layout()
plt.show()

print("The decision boundary is non-linear")
```

```python
# Train the same two-layer network (Algorithm 25, batch) on XOR: with and without biases
def train_two_layer(X, y, K=2, eta=0.1, max_iter=1000, use_bias=True, seed=0):
    rng = np.random.default_rng(seed)
    X_in = np.column_stack([X, np.ones(len(X))]) if use_bias else X
    W = rng.normal(0, 0.5, (X_in.shape[1], K))
    v = rng.normal(0, 0.5, K + int(use_bias))
    losses = []
    for it in range(max_iter):
        G, g, total = np.zeros_like(W), np.zeros_like(v), 0.0
        for x, t in zip(X_in, y):
            a = W.T @ x
            h = np.tanh(a)
            h_out = np.append(h, 1) if use_bias else h
            e = t - v @ h_out
            total += 0.5 * e**2
            g -= e * h_out
            for i in range(K):
                G[:, i] -= e * v[i] * (1 - np.tanh(a[i])**2) * x
        W -= eta * G
        v -= eta * g
        losses.append(total)
    preds = [v @ (np.append(np.tanh(W.T @ x), 1) if use_bias else np.tanh(W.T @ x)) for x in X_in]
    return W, v, losses, np.array(preds)

X_xor = np.array([[-1, -1], [-1, 1], [1, -1], [1, 1]])
y_xor = np.array([-1, 1, 1, -1])

fig, ax = plt.subplots(figsize=(11, 4.5))
for use_bias in [True, False]:
    W_x, v_x, losses, preds = train_two_layer(X_xor, y_xor, use_bias=use_bias, seed=0, max_iter=300)
    ax.plot(losses, lw=2, label=f'use_bias={use_bias}')
    print(f"use_bias={use_bias}: predictions {np.round(preds, 2)}  (targets {y_xor}),  final loss {losses[-1]:.3f}")
ax.set(xlabel='Iteration', ylabel='Loss  ½Σe²', title='Two-layer network on XOR (±1 inputs)')
ax.legend()
plt.tight_layout()
plt.show()

print("Without biases the network outputs 0 for every input. tanh is odd, so with no biases")
print("f(-x) = -f(x); but XOR needs f(1,1) = f(-1,-1) = -1. Gradient descent settles at f = 0.")

solved = sum(np.all(np.sign(train_two_layer(X_xor, y_xor, seed=s)[3]) == y_xor) for s in range(10))
print(f"\nWith biases, {solved} of 10 random initialisations solve XOR; the rest get stuck (sensitive to initialisation).")
```
